Aplicamos los 6 métodos de sustitución de valores nulos a 7 bases de Excel distintas
(media, mediana, número concreto, string concreto, forward fill y backward fill),
con su descripción por columnas, y exportamos un CSV limpio por cada base.

```python
import pandas as pd
import numpy as np

tla = pd.read_excel('Base_de_Tlaxcala.xlsx')
# Limpieza de nombres de columna (espacios invisibles)
tla.columns = tla.columns.astype(str).str.strip()
print(f"Filas: {tla.shape[0]} | Columnas: {tla.shape[1]}")
tla.isnull().sum()
```

```python
# CUARTO MÉTODO (string concreto): Telefono y Correo están vacíos al 100%,
# no hay de dónde rescatarlos -> se etiquetan
tla['Telefono'] = tla['Telefono'].fillna('Sin registro')
tla['Correo'] = tla['Correo'].fillna('Sin registro')

# QUINTO MÉTODO (forward fill): Producto solo tiene 1 hueco a media tabla;
# el producto de interés suele repetirse del registro de arriba
tla['Producto'] = tla['Producto'].ffill()

# Corroboramos y exportamos
print(f"Nulos restantes: {tla.isnull().sum().sum()}")
tla.to_csv('Base_de_Tlaxcala_sin_nulos.csv', index=False)
```

## 2. Base_de_Leads (210 registros × 14 columnas)
Base general de leads. Misma situación que Tlaxcala: **Telefono** y **Correo** vacíos
al 100%, y **Producto** con 14 huecos dispersos por toda la tabla.

```python
leads = pd.read_excel('Base_de_Leads.xlsx')
leads.columns = leads.columns.astype(str).str.strip()
print(f"Filas: {leads.shape[0]} | Columnas: {leads.shape[1]}")
leads.isnull().sum()
```

```python
# CUARTO MÉTODO (string concreto): columnas vacías al 100% -> etiqueta
leads['Telefono'] = leads['Telefono'].fillna('Sin registro')
leads['Correo'] = leads['Correo'].fillna('Sin registro')

# SEXTO MÉTODO (backward fill): los 14 huecos de Producto están dispersos;
# aquí rellenamos con el valor del registro de ABAJO (igual de válido que el de arriba)
leads['Producto'] = leads['Producto'].bfill()

# Corroboramos y exportamos
print(f"Nulos restantes: {leads.isnull().sum().sum()}")
leads.to_csv('Base_de_Leads_sin_nulos.csv', index=False)
```
